```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

temperature = np.array([20, 23, 25, 28, 30, 32, 35, 38]).reshape(-1, 1)
consumption = np.array([18, 20, 24, 29, 33, 39, 47, 49])

model = LinearRegression()
model.fit(temperature, consumption)
new_temp = np.array([[34]])
prediction = model.predict(new_temp)

print("Temperature :", new_temp[0][0], "°C")
print("Predicted Electricity Consumption :", round(prediction[0], 2), "kWh")
print("\nSlope (m):", round(model.coef_[0], 2))
print("Intercept (b):", round(model.intercept_, 2))

plt.scatter(temperature, consumption, color="blue", label="Actual Data")
plt.plot(temperature, model.predict(temperature), color="red", label="Regression Line")
plt.title("Electricity Consumption Prediction")
plt.xlabel("Temperature (°C)")
plt.ylabel("Consumption (kWh)")
plt.legend()
plt.grid(True)
plt.show()
```
